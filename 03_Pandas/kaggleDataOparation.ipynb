```python
# Extracting the text inside parentheses from the Rating column

import pandas as pd

# Load the dataset
data = pd.read_csv(r"C:\Users\ASUS\OneDrive\Desktop\Python-Data-Analysis\07_Datasets\anime-dataset-2023.csv")

# Extract text inside parentheses
def extract_data(data01):
    check = False
    st = ""

    for i in data01:
        if i == "(":
            check = True
        elif i == ")":
            return st
        elif check:
            st += i

    return st

# Apply function
data["Rating_Description"] = data["Rating"].apply(extract_data)

# Display results
print(data[["Rating", "Rating_Description"]])
```
